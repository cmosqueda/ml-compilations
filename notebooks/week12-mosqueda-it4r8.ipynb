```python
# import necessary libraries
import pdfplumber
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
# download tokenizer data -> only need to run once
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
text = ""
with pdfplumber.open("../datasets/Fin_labExer1_dataset.pdf") as pdf:
    for page in pdf.pages[15:46]:
        text += page.extract_text()

text
```

```python
# word-level tokenization
words = word_tokenize(text)
words[:30]
```

```python
# sentence-level tokenization
sentences = sent_tokenize(text)
sentences[:5]
```

```python
# display number of words and sentences
print(f"Total number of words: {len(words)}")
print(f"Total number of sentences: {len(sentences)}")
```

```python
# paragraph-level tokenization

# clean text, replace single line breaks within paragraphs
cleaned_text = text.replace('\n\n', ' ')

# split using double line breaks
paragraphs = [p.strip() for p in cleaned_text.split('\n') if p.strip()]


print("Number of paragraphs:", len(paragraphs))

print("\nDisplay first 30 paragraphs:")
for i, para in enumerate(paragraphs[:30]):
    print(f"\nParagraph {i+1}:\n{para}")
```

## Insight

I have three key takeaways for this activity.

1. Tokenizing in word and sentence levels are fairly easy because of the built in methods provided by `nltk` library.
2. It can become tricky when tokenizing unstructured data like texts, especially if its extracted from external resource via text extraction or scraping. The steps involve different preprocessing techniques like cleaning unimportant characters to only extract meaningful data and splitting data using delimiters.
3. PDFs that are not properly formatted can become difficult to extract data from.
